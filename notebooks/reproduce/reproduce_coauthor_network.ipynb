# 再現の確認（coauthor_network）

`src/coauthor_network/README.md` の工程 1〜3 を順に流し、`data/derived/coauthor_network/` の全ファイルがコミットされたものとバイト単位で一致するかを確かめる。

- どこで開いてもよい。リポジトリの一番上を探し、スクリプトはそこから流す。
- `check_orcid_names.py`（外部の API を引く）は流さない。その出力 `data/review/orcid_name_mismatches.csv` を入力として読む。
- 流す前に作業ツリーの `data/derived/coauthor_network/` がコミットされた状態であることを前提にする（流す前のハッシュを基準にする）。

```python
import hashlib
import subprocess
import sys
from pathlib import Path

# notebooks/reproduce/ で開いても動くよう、上にたどってリポジトリの一番上を探す
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents]
            if (p / "src" / "coauthor_network" / "identify_authors.py").exists())
DERIVED = ROOT / "data" / "derived" / "coauthor_network"

STEPS = [
    "identify_authors",
    "compute_team_familiarity",
    "regress_novelty_on_familiarity",
]


def snapshot() -> dict[str, str]:
    return {str(p.relative_to(ROOT)): hashlib.sha256(p.read_bytes()).hexdigest()
            for p in sorted(DERIVED.rglob("*")) if p.is_file() and p.name != ".DS_Store"}


before = snapshot()
print(f"流す前: {len(before)} ファイル")
```

```python
for i, name in enumerate(STEPS, start=1):
    print(f"[{i}] {name}")
    res = subprocess.run([sys.executable, f"src/coauthor_network/{name}.py"], cwd=ROOT,
                         capture_output=True, text=True)
    print(res.stdout)
    if res.returncode != 0:
        print(res.stderr)
        raise RuntimeError(f"工程{i} {name} が失敗した")
```

```python
after = snapshot()
changed = sorted(k for k in before.keys() & after.keys() if before[k] != after[k])
added = sorted(after.keys() - before.keys())
removed = sorted(before.keys() - after.keys())
for label, items in [("変わった", changed), ("増えた", added), ("消えた", removed)]:
    for k in items:
        print(f"{label}: {k}")
print(f"一致 {len(before.keys() & after.keys()) - len(changed)} / {len(after)} ファイル")
assert not (changed or added or removed), "コミットされた出力と一致しない"
print("✅ すべての出力がコミットされたものと一致した")
```
